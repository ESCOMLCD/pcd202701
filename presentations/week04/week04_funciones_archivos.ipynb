## Sesión 1 — Funciones

### ¿Por qué usar funciones?

Sin funciones, terminas copiando y pegando el mismo bloque de código varias veces. Cuando encuentras un bug, tienes que corregirlo en 5 lugares distintos. Las funciones resuelven eso:

- **Reutilización:** escribes el código una vez, lo usas muchas veces
- **Organización:** divides un programa grande en piezas pequeñas y manejables
- **Legibilidad:** `calcular_promedio(datos)` dice más que 5 líneas de código sueltas
- **Testing:** puedes probar cada función por separado

### Definir y llamar una función

```python
# Definir una función con def
def saludar():
    print("Hola, bienvenido a PCD!")

# Llamar la función
saludar()
saludar()    # puedes llamarla cuantas veces quieras
```

```python
# Función con parámetros
def saludar_persona(nombre):
    print(f"Hola, {nombre}!")

saludar_persona("Ana")
saludar_persona("Luis")
```

```python
# Función con return — devuelve un valor
def sumar(a, b):
    resultado = a + b
    return resultado

# El valor retornado se puede guardar en una variable
total = sumar(10, 25)
print(f"10 + 25 = {total}")

# O usarlo directamente en una expresión
print(f"Doble de la suma: {sumar(10, 25) * 2}")
```

> **`print()` vs `return`:** `print()` muestra algo en pantalla pero no devuelve nada. `return` devuelve un valor que puedes guardar y usar después. En ciencia de datos casi siempre quieres `return`, no `print()`.

```python
# Diferencia entre print y return
def con_print(a, b):
    print(a + b)       # muestra en pantalla, pero no devuelve nada

def con_return(a, b):
    return a + b       # devuelve el valor

resultado1 = con_print(3, 4)   # imprime 7
print(f"con_print retornó: {resultado1}")    # None — no devolvió nada

resultado2 = con_return(3, 4)  # no imprime nada
print(f"con_return retornó: {resultado2}")   # 7 — sí devolvió
```

### Parámetros posicionales y con nombre

```python
# Parámetros posicionales — el orden importa
def presentar(nombre, edad, carrera):
    return f"{nombre} tiene {edad} años y estudia {carrera}"

print(presentar("Ana", 21, "Ingeniería"))

# ¿Qué pasa si cambias el orden?
print(presentar(21, "Ingeniería", "Ana"))   # funciona pero el resultado no tiene sentido
```

```python
# Parámetros con nombre (keyword arguments) — el orden no importa
print(presentar(nombre="Ana", edad=21, carrera="Ingeniería"))
print(presentar(carrera="Ingeniería", nombre="Ana", edad=21))   # mismo resultado
```

```python
# Valores por defecto — se usan si no pasas ese argumento
def calcular_iva(monto, tasa=0.16):
    return monto * (1 + tasa)

print(f"Con IVA 16%: ${calcular_iva(1000):.2f}")        # usa el default 0.16
print(f"Con IVA 8%:  ${calcular_iva(1000, 0.08):.2f}")  # sobreescribe el default
```

### Retornar múltiples valores

```python
# Una función puede retornar varios valores (como una tupla)
def estadisticas(datos):
    minimo = min(datos)
    maximo = max(datos)
    promedio = sum(datos) / len(datos)
    return minimo, maximo, promedio

ventas = [1200, 850, 1500, 920, 1100, 2000, 750]

# Desempaquetar los valores retornados
mn, mx, prom = estadisticas(ventas)
print(f"Mínimo: ${mn:,}")
print(f"Máximo: ${mx:,}")
print(f"Promedio: ${prom:,.2f}")
```

### Scope: variables locales y globales

Las variables creadas **dentro** de una función son **locales** — solo existen mientras la función se ejecuta. Las variables creadas **fuera** de cualquier función son **globales**.

```python
# Variable global
mensaje_global = "Soy global"

def mi_funcion():
    # Variable local — solo existe dentro de esta función
    mensaje_local = "Soy local"
    print(f"Dentro de la función: {mensaje_local}")
    print(f"Dentro de la función: {mensaje_global}")   # puede leer la global

mi_funcion()

print(f"Fuera de la función: {mensaje_global}")   # OK

try:
    print(f"Fuera de la función: {mensaje_local}")  # ERROR — no existe aquí
except NameError as e:
    print(f"Error: {e}")
```

```python
# Cuidado: si creas una variable local con el mismo nombre que una global,
# la local "oculta" a la global dentro de la función
x = 100    # global

def cambiar_x():
    x = 999    # esta es una variable LOCAL nueva, no modifica la global
    print(f"Dentro: x = {x}")    # 999

cambiar_x()
print(f"Fuera: x = {x}")        # 100 — la global no cambió
```

### Funciones útiles para ciencia de datos

Vamos a construir funciones que te servirán directamente en la Práctica 1.

```python
# Función para contar frecuencias (lo que hicimos en semana 3, ahora reutilizable)
def contar_frecuencias(lista):
    """Cuenta cuántas veces aparece cada valor en una lista.
    Retorna un diccionario {valor: conteo}."""
    conteo = {}
    for valor in lista:
        if valor.strip() == "":    # ignorar vacíos
            continue
        conteo[valor] = conteo.get(valor, 0) + 1
    return conteo

# Probar
categorias = ["Laptop", "Mouse", "Laptop", "", "Teclado", "Mouse", "Laptop", ""]
resultado = contar_frecuencias(categorias)
print(resultado)
```

```python
# Función para encontrar el valor más frecuente
def mas_frecuente(conteo):
    """Dado un diccionario {valor: conteo}, retorna (valor, conteo) del más frecuente."""
    valor = max(conteo, key=conteo.get)
    return valor, conteo[valor]

nombre, veces = mas_frecuente(resultado)
print(f"Más frecuente: {nombre} ({veces} veces)")
```

```python
# Función para calcular estadísticas de una lista de números (como strings)
def estadisticas_columna(valores_str):
    """Recibe una lista de strings, convierte a float los que puede,
    y retorna un diccionario con min, max, promedio y n_validos."""
    numeros = []
    for v in valores_str:
        try:
            numeros.append(float(v))
        except (ValueError, TypeError):
            continue    # ignorar valores que no se pueden convertir
    
    if not numeros:
        return {"n_validos": 0, "minimo": None, "maximo": None, "promedio": None}
    
    return {
        "n_validos": len(numeros),
        "minimo": min(numeros),
        "maximo": max(numeros),
        "promedio": sum(numeros) / len(numeros)
    }

# Probar con datos que incluyen vacíos y valores no numéricos
datos = ["850.50", "1200", "", "920.30", "abc", "1500", "", "750"]
stats = estadisticas_columna(datos)
print(f"Valores válidos: {stats['n_validos']}")
print(f"Mínimo: {stats['minimo']}")
print(f"Máximo: {stats['maximo']}")
print(f"Promedio: {stats['promedio']:.2f}")
```

```python
# Función para contar celdas vacías por columna
def contar_vacios(filas, encabezados):
    """Cuenta celdas vacías por columna.
    filas: lista de listas (cada sublista es una fila)
    encabezados: lista con los nombres de las columnas
    Retorna un diccionario {columna: n_vacios}."""
    vacios = {col: 0 for col in encabezados}
    for fila in filas:
        for i, valor in enumerate(fila):
            if valor.strip() == "":
                vacios[encabezados[i]] += 1
    return vacios

# Probar
encabezados = ["nombre", "edad", "ciudad"]
filas = [
    ["Ana", "25", "CDMX"],
    ["Luis", "", "Monterrey"],
    ["", "30", ""],
    ["María", "22", "Puebla"],
]

resultado = contar_vacios(filas, encabezados)
for col, n in resultado.items():
    print(f"  {col}: {n} vacíos")
```

### `open()` y `with`

Para leer o escribir archivos en Python usas la función `open()`. La forma recomendada es con `with`, que se encarga de cerrar el archivo automáticamente cuando terminas.

```python
%%writefile ejemplo.txt
Primera linea del archivo
Segunda linea
Tercera linea
Cuarta linea
Quinta linea
```

```python
# Leer todo el archivo de una vez
with open("ejemplo.txt", "r") as archivo:
    contenido = archivo.read()

print(contenido)
print(f"Tipo: {type(contenido)}")   # es un solo string con todo el contenido
```

```python
# Leer línea por línea — la forma más común para archivos grandes
with open("ejemplo.txt", "r") as archivo:
    for numero, linea in enumerate(archivo, 1):
        print(f"Línea {numero}: {linea.strip()}")   # .strip() quita el \n al final
```

```python
# Leer todas las líneas como una lista
with open("ejemplo.txt", "r") as archivo:
    lineas = archivo.readlines()

print(f"Total de líneas: {len(lineas)}")
print(f"Primera línea: {lineas[0].strip()}")
print(f"Última línea: {lineas[-1].strip()}")
```

### Modos de apertura

| Modo | Significado | Si el archivo no existe |
|---|---|---|
| `"r"` | Lectura (read) — solo leer | Error |
| `"w"` | Escritura (write) — sobreescribe todo | Lo crea |
| `"a"` | Agregar (append) — escribe al final | Lo crea |
| `"r+"` | Lectura + escritura | Error |

```python
# Escribir un archivo (modo "w" — sobreescribe si ya existe)
with open("salida.txt", "w") as archivo:
    archivo.write("=== Reporte de datos ===\n")
    archivo.write(f"Fecha: septiembre 2026\n")
    archivo.write(f"Total de registros: 1000\n")

# Verificar
with open("salida.txt", "r") as archivo:
    print(archivo.read())
```

```python
# Agregar al final (modo "a" — no sobreescribe)
with open("salida.txt", "a") as archivo:
    archivo.write("\n--- Agregado después ---\n")
    archivo.write("Esta línea se agregó al final\n")

with open("salida.txt", "r") as archivo:
    print(archivo.read())
```

### Leer un archivo CSV sin librerías

Un archivo CSV (Comma-Separated Values) es un archivo de texto donde cada línea es una fila y los valores se separan por comas. La primera línea son los encabezados.

Vamos a crear un CSV de ejemplo y leerlo con Python puro — **esto es exactamente lo que harán en la Práctica 1**.

```python
%%writefile datos_ejemplo.csv
producto,monto_venta,email,telefono
Laptop,15000.50,ana@gmail.com,55-1234-5678
Mouse,350,luis@hotmail.com,55-8765-4321
Laptop,14500,carlosgmail.com,55-1111-2222
Teclado,800,,55-3333-4444
Monitor,5000,sofia@yahoo.com,
Laptop,,diego@@live.com,55-5555-6666
Mouse,400,val@outlook.com,5577778888
Impresora,3500,andrea@gmail.com,55-9999-0000
Mouse,320,miguel@,+52 55-1234-5678
Laptop,16000,paula@gmail.com,55-2222-3333
```

```python
# Paso 1: Leer el CSV y separar encabezados de datos
with open("datos_ejemplo.csv", "r") as archivo:
    lineas = archivo.readlines()

# La primera línea son los encabezados
encabezados = lineas[0].strip().split(",")
print(f"Encabezados: {encabezados}")
print(f"Número de columnas: {len(encabezados)}")

# Las demás líneas son datos
datos = []
for linea in lineas[1:]:
    fila = linea.strip().split(",")
    datos.append(fila)

print(f"Número de filas de datos: {len(datos)}")
print(f"Primera fila: {datos[0]}")
```

```python
# Paso 2: Imprimir las primeras 5 filas con formato
print(" | ".join(encabezados))
print("-" * 60)
for fila in datos[:5]:
    print(" | ".join(fila))
```

```python
# Paso 3: Acceder a una columna específica por nombre
# Necesitamos saber el índice de la columna
idx_producto = encabezados.index("producto")
idx_monto = encabezados.index("monto_venta")

print(f"Índice de 'producto': {idx_producto}")
print(f"Índice de 'monto_venta': {idx_monto}")

# Extraer todos los valores de la columna 'producto'
productos = [fila[idx_producto] for fila in datos]
print(f"\nProductos: {productos}")
```

```python
# Paso 4: Contar frecuencias de la columna categórica
# Reutilizamos la función que definimos antes
conteo = contar_frecuencias(productos)
print("Conteo por producto:")
for prod, n in conteo.items():
    print(f"  {prod}: {n}")

nombre, veces = mas_frecuente(conteo)
print(f"\nMás frecuente: {nombre} ({veces} veces)")
```

```python
# Paso 5: Estadísticas de la columna numérica
montos_str = [fila[idx_monto] for fila in datos]
print(f"Montos como strings: {montos_str}")
# Nota: hay valores vacíos ("") — por eso usamos nuestra función que los maneja

stats = estadisticas_columna(montos_str)
print(f"\nEstadísticas de monto_venta:")
print(f"  Valores válidos: {stats['n_validos']}")
print(f"  Mínimo: ${stats['minimo']:,.2f}")
print(f"  Máximo: ${stats['maximo']:,.2f}")
print(f"  Promedio: ${stats['promedio']:,.2f}")
```

```python
# Paso 6: Contar celdas vacías
vacios = contar_vacios(datos, encabezados)
total_vacios = sum(vacios.values())

print(f"Celdas vacías totales: {total_vacios}")
print(f"Celdas vacías por columna:")
for col, n in vacios.items():
    print(f"  {col}: {n}")
```

### Ejercicio integrador: script completo de resumen

Vamos a juntar todo en un script `.py` que lee un CSV y genera un resumen — esto es esencialmente la Práctica 1.

```python
%%writefile resumen_csv.py
"""Script que lee un CSV y genera un resumen en texto.
Este es un ejemplo simplificado de lo que harán en la Práctica 1."""


def leer_csv(ruta):
    """Lee un CSV y retorna (encabezados, datos).
    encabezados: lista de strings
    datos: lista de listas (cada sublista es una fila)"""
    with open(ruta, "r", encoding="utf-8") as archivo:
        lineas = archivo.readlines()
    encabezados = lineas[0].strip().split(",")
    datos = [linea.strip().split(",") for linea in lineas[1:]]
    return encabezados, datos


def extraer_columna(datos, encabezados, nombre_columna):
    """Extrae todos los valores de una columna por su nombre."""
    idx = encabezados.index(nombre_columna)
    return [fila[idx] for fila in datos]


def contar_frecuencias(lista):
    """Cuenta frecuencias ignorando vacíos."""
    conteo = {}
    for valor in lista:
        if valor.strip() == "":
            continue
        conteo[valor] = conteo.get(valor, 0) + 1
    return conteo


def estadisticas_numerica(valores_str):
    """Calcula min, max, promedio de una lista de strings numéricos."""
    numeros = []
    for v in valores_str:
        try:
            numeros.append(float(v))
        except (ValueError, TypeError):
            continue
    if not numeros:
        return {"n_validos": 0, "minimo": None, "maximo": None, "promedio": None}
    return {
        "n_validos": len(numeros),
        "minimo": min(numeros),
        "maximo": max(numeros),
        "promedio": sum(numeros) / len(numeros)
    }


def contar_vacios(datos, encabezados):
    """Cuenta celdas vacías por columna."""
    vacios = {col: 0 for col in encabezados}
    for fila in datos:
        for i, valor in enumerate(fila):
            if valor.strip() == "":
                vacios[encabezados[i]] += 1
    return vacios


def generar_resumen(ruta_csv, col_categorica, col_numerica, ruta_salida):
    """Lee un CSV y genera un archivo de resumen."""
    # Leer datos
    encabezados, datos = leer_csv(ruta_csv)
    
    # Análisis
    categorias = extraer_columna(datos, encabezados, col_categorica)
    conteo = contar_frecuencias(categorias)
    valor_top = max(conteo, key=conteo.get)
    
    valores_num = extraer_columna(datos, encabezados, col_numerica)
    stats = estadisticas_numerica(valores_num)
    
    vacios = contar_vacios(datos, encabezados)
    total_vacios = sum(vacios.values())
    
    # Escribir resumen
    with open(ruta_salida, "w", encoding="utf-8") as f:
        f.write(f"=== RESUMEN DEL DATASET ===\n")
        f.write(f"Archivo: {ruta_csv}\n")
        f.write(f"\n--- Dimensiones ---\n")
        f.write(f"Filas: {len(datos)}\n")
        f.write(f"Columnas: {len(encabezados)}\n")
        f.write(f"Nombres de columnas: {', '.join(encabezados)}\n")
        f.write(f"\n--- Primeras 5 filas ---\n")
        f.write(" | ".join(encabezados) + "\n")
        for fila in datos[:5]:
            f.write(" | ".join(fila) + "\n")
        f.write(f"\n--- Columna categórica: {col_categorica} ---\n")
        f.write(f"Valores únicos: {len(conteo)}\n")
        f.write(f"Valor más frecuente: {valor_top} ({conteo[valor_top]} apariciones)\n")
        f.write(f"\n--- Columna numérica: {col_numerica} ---\n")
        f.write(f"Valores válidos (no vacíos): {stats['n_validos']}\n")
        f.write(f"Mínimo: {stats['minimo']}\n")
        f.write(f"Máximo: {stats['maximo']}\n")
        f.write(f"\n--- Calidad de datos ---\n")
        f.write(f"Celdas vacías totales: {total_vacios}\n")
        f.write(f"Celdas vacías por columna:\n")
        for col, n in vacios.items():
            f.write(f"  {col}: {n}\n")
    
    print(f"Resumen generado en: {ruta_salida}")


# Ejecutar
generar_resumen(
    ruta_csv="datos_ejemplo.csv",
    col_categorica="producto",
    col_numerica="monto_venta",
    ruta_salida="resumen_generado.txt"
)
```
